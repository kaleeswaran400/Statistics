ANOVA Analysis

```python
independent_var=['gender','item_purchased','category','location','size','color','season','subscription_status','shipping_type','discount_applied','payment_method','previous_purchases_category']
dependent_var=['purchase_amount','review_rating','purchase_frequency']
```

```python
result_table=[]
factors= itertools.combinations(independent_var,2)
i='purchase_amount'
for j in factors:
        model=ols(f'{i}~C({j[0]})+C({j[1]})+C({j[0]}):C({j[1]})', data=campaign_data).fit()
        result=sm.stats.anova_lm(model, typ=2)
        for k in result.index:
            if k != "Residual":
                result_table.append(pd.DataFrame([{"Dependent_variable":i,"Factor_1":k,
                                    "DF":result.loc[k,"df"],
                                    "F":result.loc[k,"F"],
                                    "P_Value":result.loc[k,"PR(>F)"]}]))
Final_Aova1=pd.concat(result_table, ignore_index=True)

```

```python
result_table=[]
factors= itertools.combinations(independent_var,2)
i='review_rating'
campaign_data1=campaign_data.dropna()
for j in factors:
        model=ols(f'{i}~C({j[0]})+C({j[1]})+C({j[0]}):C({j[1]})', data=campaign_data1).fit()
        result=sm.stats.anova_lm(model, typ=2)
        for k in result.index:
            if k != "Residual":
                result_table.append(pd.DataFrame([{"Dependent_variable":i,"Factor_1":k,
                                    "DF":result.loc[k,"df"],
                                    "F":result.loc[k,"F"],
                                    "P_Value":result.loc[k,"PR(>F)"]}]))
Final_Aova2=pd.concat(result_table, ignore_index=True)
```

```python
result_table=[]
factors= itertools.combinations(independent_var,2)
i='purchase_frequency'
for j in factors:
        model=ols(f'{i}~C({j[0]})+C({j[1]})+C({j[0]}):C({j[1]})', data=campaign_data).fit()
        result=sm.stats.anova_lm(model, typ=2)
        for k in result.index:
            if k != "Residual":
                result_table.append(pd.DataFrame([{"Dependent_variable":i,"Factor_1":k,
                                    "DF":result.loc[k,"df"],
                                    "F":result.loc[k,"F"],
                                    "P_Value":result.loc[k,"PR(>F)"]}]))
Final_Aova3=pd.concat(result_table, ignore_index=True)
```

```python
Final_Aova1["Significance"]=Final_Aova1["P_Value"].apply(lambda x : "True" if x<0.05 else "False")
print(Final_Aova1.shape)
```

```python
Final_Aova2["Significance"]=Final_Aova2["P_Value"].apply(lambda x : "True" if x<0.05 else "False")
print(Final_Aova2.shape)
```

```python
Final_Aova3["Significance"]=Final_Aova3["P_Value"].apply(lambda x : "True" if x<0.05 else "False")
print(Final_Aova3.shape)
```

```python
for table_name, table in [('Purchase amount', Final_Aova1), ('Review rating', Final_Aova2), ('Purchase frequency', Final_Aova3)]:
    significant = table[table['P_Value'] < 0.05].sort_values('P_Value')
    print(f'\n{table_name}: {len(significant)} significant effects out of {len(table)} tests')
    print(significant[['Factor_1', 'F', 'P_Value']].to_string(index=False))
```

# Statistical Test Conclusion

## Test design

Three two-factor ANOVA result tables were produced for the customer-shopping data:

- `Final_Aova1`: purchase amount
- `Final_Aova2`: review rating
- `Final_Aova3`: purchase frequency, represented numerically as approximate days between purchases

For every pair of the 12 independent variables, the model tested both main effects and their interaction. Each table contains 198 tested terms. At the 5% significance level, an effect was treated as statistically significant when `P_Value < 0.05`; otherwise, there was insufficient evidence to reject the null hypothesis. The significance columns now correctly use the p-values from their own ANOVA table.

## ANOVA results

- **Purchase amount:** 21 of 198 terms were significant. The strongest recurring evidence involved `location` (`F = 3.2943`, `p < 0.001`), followed by effects involving `season` and `previous_purchases_category` (`p` values approximately 0.0048 to 0.0376). This indicates that spending varies across some locations, seasons, and loyalty groups, although the ANOVA table alone does not identify the specific levels that differ.
- **Review rating:** 24 of 198 terms were significant. `location` was the strongest result (`F = 6.5463`, `p < 0.001`). There was also evidence involving `size` and `color`, plus interactions between `category` and `payment_method`, and between `discount_applied` and `previous_purchases_category`. This suggests that ratings are associated with both product/customer grouping and some combined conditions, rather than one universal rating pattern.
- **Purchase frequency:** 4 of 198 terms were significant. The significant results were interactions between `shipping_type` and `previous_purchases_category` (`p = 0.0214`), `season` and `shipping_type` (`p = 0.0290`), `item_purchased` and `color` (`p = 0.0334`), and `size` and `discount_applied` (`p = 0.0490`). The interaction terms mean that the relationship with purchase frequency depends on the other factor; they should not be interpreted as standalone effects.

The ANOVA results show evidence of differences, but they do not specify which category levels differ. Because many terms were tested, individual p-values near 0.05 should be treated cautiously and supported with effect sizes, group means, and sample sizes.

## Tukey HSD conclusion

Tukey HSD was used as the post-hoc test at alpha = 0.05. The displayed filtered result contains three statistically significant pairwise differences:

- Purchase frequency differs between the `Green` and `Turquoise` colour groups (`p-adj = 0.0345`). The negative mean difference indicates that Green has a lower encoded frequency value than Turquoise. Since frequency labels were mapped to days, this is a difference in the encoded purchase interval, not automatically a difference in purchase count.
- Purchase frequency differs between `Fall` and `Spring` (`p-adj = 0.0407`), with Fall lower by approximately 2.82 encoded days.
- Purchase frequency differs between `Fall` and `Summer` (`p-adj = 0.0182`), with Fall lower by approximately 3.15 encoded days.

These Tukey results identify specific group pairs behind detectable frequency differences. No additional pairwise differences should be claimed from the filtered output. A non-significant comparison means that the data do not provide sufficient evidence of a difference at the 5% level; it does not prove that the group means are identical.

## Overall conclusion

Taken together, the tests indicate that customer spending, review ratings, and purchase frequency are not completely uniform across the observed customer, product, and transaction groups. Spending shows repeated differences associated with location, season, and previous-purchase segment. Ratings show strong location differences and additional size, colour, and interaction effects. Frequency has fewer omnibus findings, but its significant interactions show that purchasing intervals depend on combinations such as shipping type with loyalty segment or season.

The post-hoc evidence provides the clearest level-specific frequency findings: Fall has a shorter encoded purchase interval than Spring and Summer, and Green has a shorter encoded interval than Turquoise. These findings can guide further investigation of seasonal planning, colour-level merchandising, shipping experience, and loyalty-focused campaigns.

The results are associative, not causal. The data are observational, the frequency labels were converted into approximate day values, and unmeasured customer or product differences may explain part of the observed variation. Before using the results for production decisions, remove the five trailing artifact rows identified in the project report, use `.dropna()` consistently for both variables in every Tukey comparison, review group sizes and effect sizes, and validate promising interventions with dated transaction data or a controlled experiment.

```python
turkey=[]
from statsmodels.stats.multicomp import pairwise_tukeyhsd

for i in dependent_var:
    for j in independent_var:
        print(campaign_data[[i,j]].dropna().dtypes)
        result=pairwise_tukeyhsd(endog=campaign_data[i], groups=campaign_data[j], alpha=0.05)
        result1=pd.DataFrame(result.summary().data[1:], columns=result.summary().data[0])
        result1.insert(0,"dependent_var",i)
        result1.insert(1,"Independent_var" ,j)
        turkey.append(result1)

final_tukey=pd.concat(turkey, ignore_index=True)

final_tukey["Significant"] = final_tukey["p-adj"].apply(lambda x: "Yes" if x < 0.05 else "No")  
```

# T Test

***One same T test***

** Is the average CPA significantly different from 1.672?**

```python

success='''Null hypotehsis:
    there is no difference between the bench mark CPC, which is 1.66 and the actual CPC'''
Failure='''Alternate Hypothesis
    There is  difference between the bench mark CPC, which is 1.66 and the actual CPC
 '''
campaign_data["CPC"].mean()

from scipy.stats import ttest_1samp

stats,p= ttest_1samp(campaign_data["CPC"].dropna(), popmean=1.672)
print("Two sided T test")
print()
print(f"calculate t_value={stats}, and p value is {p}")
print()

if p<0.05:
    print(Failure)
else:
    print(success)

success='''Null hypotehsis:
 Bench mark CPC, which is 1.66 > the actual CPC'''
Failure='''Alternate Hypothesis
Bench mark CPC, which is 1.66 < the actual CPC''' 

stats,p= ttest_1samp(campaign_data["CPC"].dropna(), popmean=1.672, alternative="less")
print("Left sided T test")
print()
print(f"calculate t_value={stats}, and p value is {p}")
print()
if p<0.05:
    print(Failure)
else:
    print(success)

success=''' Null hypotehsis:
Bench mark CPC, which is 1.66 < the actual CPC'''
Failure='''
 Alternate Hypothesis
Bench mark CPC, which is 1.66 > the actual CPC''' 

stats,p= ttest_1samp(campaign_data["CPC"].dropna(), popmean=1.672, alternative="greater")
print("Right sided T test")
print()
print(f"calculate t_value={stats}, and p value is {p}")
print()
if p<0.05:
    print(Failure)
else:
    print(success)
```

Conduct Tukey's HSD for each possible combination to identify the difference

```python
import pandas as pd
from statsmodels.stats.multicomp import pairwise_tukeyhsd

dependent_var = ["ROAS", "CPC", "CPA", "CTR", "revenue", "impressions"]
independent_var = ["campaign_type", "industry", "country", "platform"]

# Create an empty list to store row dictionaries
all_results = []

for dep in dependent_var:
    for indep in independent_var:
        # Drop missing values across both specific columns to avoid errors
        clean_df = campaign_data[[dep, indep]].dropna()
        
        # Ensure we have data and more than one group present to test
        if not clean_df.empty and clean_df[indep].nunique() > 1:
            # Run Tukey's HSD using the loop variables
            tukey = pairwise_tukeyhsd(endog=clean_df[dep], groups=clean_df[indep], alpha=0.05)
            
            # Convert the Tukey result table directly into a pandas DataFrame
            tukey_df = pd.DataFrame(data=tukey.summary().data[1:], columns=tukey.summary().data[0])
            
            # Add context columns so you know which variables produced these rows
            tukey_df.insert(0, "Dependent_Var", dep)
            tukey_df.insert(1, "Independent_Var", indep)
            
            all_results.append(tukey_df)

# Combine all individual dataframes into one master table
final_tukey_table = pd.concat(all_results, ignore_index=True)

# introduce a new column in the final__tukey_table to indicate significance based on p-value
final_tukey_table["Significant"] = final_tukey_table["p-adj"].apply(lambda x: "Yes" if x < 0.05 else "No")  

# Display the master table with renamed columns for clarity
print(final_tukey_table.reset_index(drop=True).rename(columns={"Dependent_Var":"Dependent Variable", "Independent_Var":"Independent Variable"}))
```

# Two ANOVA

Conduct Two_way ANOVA for each possbile dependent variable with independent variable

```python
from statsmodels.formula.api import ols
import statsmodels.api as sm
import itertools
dependent_var=["ROAS","CPC","CPA","CTR","revenue","impressions"]
independent_var=["campaign_type","industry","country","platform"]   
factor=list(itertools.combinations(independent_var,2))
 
result=[]
 
for i in dependent_var:
    for j in range(len(factor)):
        factor1,factor2=factor[j]
        model=ols(f'{i}~C({factor1})+C({factor2})+C({factor1}):C({factor2})', data=campaign_data).fit()
        ANOVA1=sm.stats.anova_lm(model,typ=2)
        for k in ANOVA1.index:
                if k!= "Residual":
                   result.append(pd.DataFrame( [{"Dependent_var": i,
                                   "Factor1":k,
                                    "Sum_sq":ANOVA1.loc[k,"sum_sq"],
                                    "DF":ANOVA1.loc[k,"df"],
                                    "F":ANOVA1.loc[k,"F"],
                                    "P_Value":ANOVA1.loc[k,"PR(>F)"]}]))
final_ANOVA1=pd.concat(result, ignore_index=True)
final_ANOVA1["Significant"] = final_ANOVA1["P_Value"].apply(lambda x: "Yes" if x < 0.05 else "No")
final_ANOVA1 = final_ANOVA1.rename(columns={"Factor1": "Independent_Var"})
final_ANOVA1["Independent_Var"] = final_ANOVA1["Independent_Var"].str.replace(r"C\(([^)]+)\)", r"\1", regex=True)
display(final_ANOVA1)
```

Extract the resule from the final anova table where the significance is True

```python
# Filter significant results from both statistical tests
Significant_ANOVA = final_ANOVA1[final_ANOVA1["P_Value"] < 0.05].copy()
Signicant_tukeyHSD = final_tukey_table[final_tukey_table["Significant"] == "Yes"].copy()
 
# Standardize both result tables before combining them
anova_summary = (
    Significant_ANOVA
    .assign(
        Test="Two-way ANOVA",
        group1=pd.NA,
        group2=pd.NA
    )
    .rename(columns={
        "Dependent_var": "Dependent_Var",
        "Factor1": "Independent_Var",
        "Sum_sq": "sum_sq",
        "DF": "df",
        "F": "F_Statistic"
    })
 )
 
tukey_summary = (
    Signicant_tukeyHSD
    .assign(Test="Tukey HSD")
    .rename(columns={"p-adj": "P_Value"})
 )
 
master_results = pd.concat(
    [anova_summary, tukey_summary],
    ignore_index=True,
    sort=False
 ).reindex(columns=[
    "Test",
    "Dependent_Var",
    "Independent_Var",
    "group1",
    "group2",
    "sum_sq",
    "df",
    "F_Statistic",
    "P_Value",
    "Significant"
 ])
 
display(master_results)
```

# Statistical conclusion:

The Two-way ANOVA shows that the advertising platform significantly affects ROAS, CPC, CPA, CTR, and revenue. Therefore, the average performance of these metrics is different across Google Ads, Meta Ads, and TikTok Ads.

The Tukey HSD test confirms that every platform pair differs significantly for these metrics:

* Google Ads vs Meta Ads
* Google Ads vs TikTok Ads
* Meta Ads vs TikTok Ads
However, impressions does not show a statistically significant difference between platforms. This suggests that the platform has a stronger effect on campaign quality and financial performance than on total reach.
